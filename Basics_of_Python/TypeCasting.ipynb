```python
#Type casting- str(), int(), float(), bool()
```

```python
name = "Animesh"
age = 24
is_student = True
gpa = 8.5
print(name, age, is_student, gpa)
type(name), type(age), type(is_student), type(gpa)
```

```python
name_boolean = bool(name) # Non empty strings are True
age_string = str(age)
age_float = float(age)
gpa_int = int(gpa)
is_student_int = int(is_student) # True = 1, False = 0
```

```python
print("name_boolean: type")
name_boolean, type(name_boolean)
```

```python
print("age_string: type")
age_string, type(age_string)
```

```python
print("age_float: type")
age_float, type(age_float)
```

```python
print("gpa_int: type")
gpa_int, type(gpa_int)
```

```python
print("is_student: type")
is_student_int, type(is_student_int)
```
